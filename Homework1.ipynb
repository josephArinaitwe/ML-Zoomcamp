```python
most_frequent = df['horsepower'].mode()[0]
df["horsepower"] = df["horsepower"].fillna(most_frequent)
```

```python
asia_cars = df[df["origin"] == "Asia"]
selected = asia_cars[["vehicle_weight", "model_year"]].head(7)
```

```python
# 4. Convert the selected rows to a NumPy array
X = selected.to_numpy()
X
```

```python
# 5. Multiply the transpose of X by X
XTX = X.T @ X

# 6. Invert the result
XTX_inv = np.linalg.inv(XTX)

# 7. Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Calculate the weights
w = XTX_inv @ X.T @ y
w
```
